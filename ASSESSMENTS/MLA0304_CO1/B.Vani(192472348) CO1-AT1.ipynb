**Experiment 3: Implementation of the Reinforcement Learning Framework**

```python
import gymnasium as gym
import random
# Create environment
env = gym.make("FrozenLake-v1", is_slippery=False)
# Number of episodes
episodes = 5
for episode in range(episodes):
    state, info = env.reset()
    done = False
    total_reward = 0
    print(f"\nEpisode {episode + 1}")
    while not done:
        # Agent selects a random action
        action = random.randint(0, env.action_space.n - 1)
        # Perform the action
        next_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated
        print(f"State: {state} | Action: {action} | Next State: {next_state} | Reward: {reward}")
        state = next_state
        total_reward += reward
    print("Total Reward:", total_reward)
env.close()
print("\nBasic Reinforcement Learning framework implemented successfully!")
```

**Experiment 4: Simulation of a Markov Decision Process (MDP)**

```python
states = ["Home", "Office", "Park"]
transitions = {
    "Home": ("Office", 5),
    "Office": ("Park", 10),
    "Park": ("Home", 2)
}
current_state = "Home"
print("Starting State:", current_state)
# Simulate 5 transitions
for step in range(5):
    next_state, reward = transitions[current_state]
    print(f"Step {step+1}")
    print("Current State :", current_state)
    print("Next State    :", next_state)
    print("Reward         :", reward)
    current_state = next_state
```

**Experiment 5: Bellman Equation Demonstration**

```python
gamma = 0.9
rewards = [5, 10]
V = [0, 0]
print("Initial State Values:", V)
for i in range(10):
    V0 = rewards[0] + gamma * V[1]
    V1 = max(
        rewards[0] + gamma * V[0],
        rewards[1] + gamma * V[0]
    )
    V = [round(V0, 2), round(V1, 2)]
    print(f"Iteration {i+1}: {V}")
print("\nFinal State Values:", V)
```

**Experiment 6: Exploration vs. Exploitation Strategies**

```python
import random
actions = ["Left", "Right", "Up", "Down"]
q_values = [5, 8, 3, 6]
epsilon = 0.2
print("Greedy Strategy")
greedy_action = actions[q_values.index(max(q_values))]
print("Selected Action:", greedy_action)
print("\nRandom Strategy")
random_action = random.choice(actions)
print("Selected Action:", random_action)
print("\nEpsilon-Greedy Strategy")
if random.random() < epsilon:
    action = random.choice(actions)
    print("Exploration")
else:
    action = actions[q_values.index(max(q_values))]
    print("Exploitation")
print("Selected Action:", action)
```

**Experiment 7: Multi-Armed Bandit Problem**

```python
import random
import math
rewards = [2, 5, 8, 4]
epsilon = 0.2
steps = 20
q = [0] * 4
count = [0] * 4
total_reward = 0
print("Epsilon-Greedy")
for i in range(steps):
    if random.random() < epsilon:
        action = random.randint(0, 3)
    else:
        action = q.index(max(q))
    reward = rewards[action] + random.randint(-1, 1)
    count[action] += 1
    q[action] += (reward - q[action]) / count[action]
    total_reward += reward
print("Cumulative Reward:", total_reward)
q = [0] * 4
count = [0] * 4
total_reward = 0
print("\nUCB")
for i in range(steps):
    if 0 in count:
        action = count.index(0)
    else:
        ucb = [
            q[j] + math.sqrt((2 * math.log(i + 1)) / count[j])
            for j in range(4)
        ]
        action = ucb.index(max(ucb))
    reward = rewards[action] + random.randint(-1, 1)
    count[action] += 1
    q[action] += (reward - q[action]) / count[action]
    total_reward += reward
print("Cumulative Reward:", total_reward)
```

**Experiment 8: Reward Visualization in Reinforcement Learning**

```python
import random
import matplotlib.pyplot as plt
episodes = 20
episode_rewards = []
cumulative_rewards = []
total_reward = 0
for i in range(episodes):
    reward = random.randint(5, 15)
    episode_rewards.append(reward)
    total_reward += reward
    cumulative_rewards.append(total_reward)
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(range(1, episodes+1), episode_rewards, marker='o')
plt.title("Episode Rewards")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.subplot(1,2,2)
plt.plot(range(1, episodes+1), cumulative_rewards, marker='o')
plt.title("Cumulative Rewards")
plt.xlabel("Episode")
plt.ylabel("Cumulative Reward")
plt.tight_layout()
plt.show()
```
